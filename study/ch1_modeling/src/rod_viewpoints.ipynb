```python
import os
import numpy as np  # type: ignore
import matplotlib.pyplot as plt  # type: ignore
from matplotlib.patches import Rectangle, Circle  # type: ignore
from datetime import datetime

RESULT_DIR = os.path.join(os.getcwd(), "result")
os.makedirs(RESULT_DIR, exist_ok=True)
TIMESTAMP = datetime.now().strftime("%Y%m%d_%H%M%S")

# ─────────────────────────────────────────────
# STEP 1: Three ways of looking at the same rod
#
# What we have: one rod, length L = 1.0 m, radius R = 0.02 m.
# What we want: to see WHERE the viewer stands in each derivation.
#
# NOT TO SCALE: the radius is drawn at 0.08 (instead of 0.02) so the rod is visible next to its length.
# ─────────────────────────────────────────────

ROD_LENGTH = 1.0  # m, foundational
DRAW_RADIUS = 0.08  # m, drawing only (the real radius is 0.02 m)

fig, axes = plt.subplots(1, 3, figsize=(14, 4.6))

# panel a: SIDE VIEW, axis perpendicular to the rod (the I_com derivation)
ax = axes[0]
ax.add_patch(Rectangle((-ROD_LENGTH / 2, -DRAW_RADIUS), ROD_LENGTH, 2 * DRAW_RADIUS, color="C0", alpha=0.4))
ax.plot([0], [0], "o", color="red", markersize=13, zorder=5, label="axis (a pin pointing at you)")
ax.annotate("you look from here\n(from the side)", xy=(0, -0.12), xytext=(0, -0.42), ha="center", arrowprops=dict(arrowstyle="->", lw=2))
ax.set_title("(a) Side view\nused for the moment of inertia about the center of mass")
ax.set_xlim(-0.7, 0.7)
ax.set_ylim(-0.55, 0.35)
ax.set_aspect("equal")
ax.set_xticks([])
ax.set_yticks([])
ax.legend(loc="upper right", fontsize=8)

# panel b: the same SIDE VIEW, but now the axis runs along the rod
ax = axes[1]
ax.add_patch(Rectangle((-ROD_LENGTH / 2, -DRAW_RADIUS), ROD_LENGTH, 2 * DRAW_RADIUS, color="C0", alpha=0.4))
ax.plot([-0.6, 0.6], [0, 0], "--", color="red", linewidth=2, label="axis (a line along the rod)")
ax.annotate("you look from here\n(from the side)", xy=(0, -0.12), xytext=(0, -0.42), ha="center", arrowprops=dict(arrowstyle="->", lw=2))
ax.set_title("(b) The same side view, but the axis runs along the rod\nfrom here the axis is a LINE, not a dot")
ax.set_xlim(-0.7, 0.7)
ax.set_ylim(-0.55, 0.35)
ax.set_aspect("equal")
ax.set_xticks([])
ax.set_yticks([])
ax.legend(loc="upper right", fontsize=8)

# panel c: END VIEW, looking along the axis (the I_zz derivation)
ax = axes[2]
ax.add_patch(Circle((0, 0), 0.5, color="C0", alpha=0.4))
ax.plot([0], [0], "o", color="red", markersize=13, zorder=5, label="axis (a pin pointing at you)")
ax.plot([0, -0.5], [0, 0], "-", color="black", linewidth=1.5)
ax.text(-0.25, 0.04, "R", ha="center")
point = np.array([0.3, 0.2])
ax.plot(*point, "o", color="C1", markersize=8)
ax.annotate("", xy=point, xytext=(0, 0), arrowprops=dict(arrowstyle="->", color="C1", lw=2))
ax.text(0.2, 0.02, "s", color="C1", fontsize=12)
ax.set_title("(c) End view, looking along the axis\nused for the moment of inertia about the long axis")
ax.set_xlabel("you look from here: the end of the rod, looking down its length")
ax.set_xlim(-0.7, 0.7)
ax.set_ylim(-0.7, 0.7)
ax.set_aspect("equal")
ax.set_xticks([])
ax.set_yticks([])
ax.legend(loc="upper right", fontsize=8)

fig.suptitle("Same rod, different place to stand (not to scale: the rod is drawn thicker than it is)")
fig.tight_layout()
fig.savefig(os.path.join(RESULT_DIR, f"viewpoint_axis_views_{TIMESTAMP}.png"))
print("saved the figure")
plt.show()
```
